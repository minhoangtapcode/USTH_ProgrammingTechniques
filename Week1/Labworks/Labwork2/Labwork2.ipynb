## Exercise 1
Add typing (parameters and return values) to the following code.
*(It is `x**y` by exponentiation-by-squaring, pure Python - no numpy needed.)*

```python
%%writefile code/exercise1.py
def power(x: float, y: int) -> float:
    result = 1.0
    while True:
        t = y % 2
        y = y // 2
        if t == 1:
            result = result * x
        if y == 0:
            break
        x = x * x
    return result


if __name__ == "__main__":
    print(f"power(2, 10)   = {power(2, 10)}")
    print(f"power(2.0, 10) = {power(2.0, 10)}")
```

## Exercise 3
Add typing to every parameter and returned value of the following linked list
(adapted from [realpython](https://realpython.com/linked-lists-python/)).
*Tip: the nested `__Node` is private and self-referential - a `from __future__ import annotations`
and a type alias help.*

```python
%%writefile code/exercise3.py
from __future__ import annotations
from typing import Iterator, Optional

class LinkedList:
    class _Node:
        def __init__(self, data: str, next: Optional[LinkedList._Node] = None) -> None:
            self.__data: str = data
            self.__next: Optional[LinkedList._Node] = next

        @property
        def data(self) -> str:
            return self.__data

        @property
        def next(self) -> Optional[LinkedList._Node]:
            return self.__next

        @next.setter
        def next(self, _next: Optional[LinkedList._Node]) -> None:
            self.__next = _next

        def __repr__(self) -> str:
            return repr(self.data)

    # Type alias help for the private, self-referential nested class
    OptNode = Optional[_Node]

    def __init__(self, nodes: Optional[list[str]] = None) -> None:
        self.__head: LinkedList.OptNode = None
        if nodes is not None and len(nodes) > 0:
            node = self._Node(data=nodes.pop(0))
            self.__head = node
            for elem in nodes:
                new_node = self._Node(elem)
                node.next = new_node
                node = new_node

    def __repr__(self) -> str:
        node: LinkedList.OptNode = self.__head
        nodes: list[str] = []
        while node is not None:
            nodes.append(repr(node))
            node = node.next
        nodes.append("None")
        return " -> ".join(nodes)

    def __iter__(self) -> Iterator[str]:
        node: LinkedList.OptNode = self.__head
        while node is not None:
            yield node.data
            node = node.next

    def is_empty(self) -> bool:
        return self.__head is None

    def add_first(self, data: str) -> LinkedList:
        self.__head = self._Node(data, self.__head)
        return self

    def add_last(self, data: str) -> LinkedList:
        # Check against self.__head directly so mypy knows it isn't None below
        if self.__head is None:
            return self.add_first(data)
            
        current_node = self.__head
        while current_node.next is not None:
            current_node = current_node.next
        current_node.next = self._Node(data)
        return self


if __name__ == "__main__":
    lklist = LinkedList(["b", "d", "f"])
    lklist.add_first("a")
    lklist.add_last("g")
    print(lklist)

       
```

## Exercise 4
Restart from Labwork 1, Exercise 4 (the multiple-inheritance scheme). Copy your solution and add all the necessary typing.

```python
%%writefile code/exercise4.py
from __future__ import annotations
from enum import Enum

# paste your Labwork 1 / Exercise 4 solution and add typing
class FuelType(Enum):
    DIESEL = "diesel"
    GASOLINE = "gasoline"
    KEROSENE = "kerosene"
    COAL = "coal"
    ELECTRICITY = "electricity"
    SOLAR = "solar"

class Vehicle:
    def __init__(self, capacity: int, is_engine_powered: bool, max_speed: int) -> None:
        self.capacity: int = capacity
        self.is_engine_powered: bool = is_engine_powered
        self.max_speed: int = max_speed

# Engine component for vehicles with an engine
class Engine:
    def __init__(self, fuel_type: FuelType, average_consumption: float) -> None:
        self.fuel_type: FuelType = fuel_type
        self.average_consumption: float = average_consumption  # e.g., liters per 100km or kWh per km

    def get_fuel_type(self) -> FuelType:
        return self.fuel_type

    def get_average_consumption(self) -> float:
        return self.average_consumption

# Vehicles without engine (inheriting only from Vehicle)
class Donkey(Vehicle):
    def __init__(self) -> None:
        super().__init__(capacity=2, is_engine_powered=False, max_speed=15)

class Bicycle(Vehicle):
    def __init__(self) -> None:
        super().__init__(capacity=1, is_engine_powered=False, max_speed=30)

# Vehicles with engine (using multiple inheritance: Vehicle + Engine)
class Boat(Vehicle, Engine):
    def __init__(self, fuel_type: FuelType, average_consumption: float) -> None:
        Vehicle.__init__(self, capacity=15, is_engine_powered=True, max_speed=60)
        Engine.__init__(self, fuel_type=fuel_type, average_consumption=average_consumption)

class Car(Vehicle, Engine):
    def __init__(self, fuel_type: FuelType, average_consumption: float) -> None:
        Vehicle.__init__(self, capacity=5, is_engine_powered=True, max_speed=200)
        Engine.__init__(self, fuel_type=fuel_type, average_consumption=average_consumption)

class Airplane(Vehicle, Engine):
    def __init__(self, fuel_type: FuelType, average_consumption: float) -> None:
        Vehicle.__init__(self, capacity=250, is_engine_powered=True, max_speed=900)
        Engine.__init__(self, fuel_type=fuel_type, average_consumption=average_consumption)

if __name__ == '__main__':
    my_car = Car(fuel_type=FuelType.GASOLINE, average_consumption=7.5)
    print(f"Car Capacity: {my_car.capacity}, Fuel: {my_car.get_fuel_type().value}, Consumption: {my_car.get_average_consumption()}L/100km")
    
    my_bike = Bicycle()
    print(f"Bicycle Capacity: {my_bike.capacity}, Engine Powered: {my_bike.is_engine_powered}")

    my_boat = Boat(fuel_type=FuelType.DIESEL, average_consumption=10.0)
    print(f"Boat Capacity: {my_boat.capacity}, Fuel: {my_boat.get_fuel_type().value}, Consumption: {my_boat.get_average_consumption()}L/100km")

    my_airplane = Airplane(fuel_type=FuelType.KEROSENE, average_consumption=2.5)
    print(f"Airplane Capacity: {my_airplane.capacity}, Fuel: {my_airplane.get_fuel_type().value}, Consumption: {my_airplane.get_average_consumption()}L/100km")
```
