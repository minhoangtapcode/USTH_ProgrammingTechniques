## Exercise 3
Write a class to represent a vehicle:
- it can transport a given number of people;
- it can be engine-powered or not;
- it has a maximum speed.

Then write classes for bicycles, cars, boats and planes.

```python
%%writefile code/exercise3.py
# write your solution here
class Vehicle:
    def __init__(self, capacity, is_engine_powered, max_speed):
        self.capacity = capacity
        self.is_engine_powered = is_engine_powered
        self.max_speed = max_speed

class Bicycle(Vehicle):
    def __init__(self):
        super().__init__(capacity=1, is_engine_powered=False, max_speed=20)

class Car(Vehicle):
    def __init__(self):
        super().__init__(capacity=5, is_engine_powered=True, max_speed=200)

class Boat(Vehicle):
    def __init__(self):
        super().__init__(capacity=15, is_engine_powered=True, max_speed=60) 

class Plane(Vehicle):
    def __init__(self):
        super().__init__(capacity=250, is_engine_powered=True, max_speed=760)       

if __name__ == '__main__':
    bicycle = Bicycle()
    car = Car()
    boat = Boat()
    plane = Plane()

    print(f"Bicycle: capacity={bicycle.capacity}, is_engine_powered={bicycle.is_engine_powered}, max_speed={bicycle.max_speed}")
    print(f"Car: capacity={car.capacity}, is_engine_powered={car.is_engine_powered}, max_speed={car.max_speed}")
    print(f"Boat: capacity={boat.capacity}, is_engine_powered={boat.is_engine_powered}, max_speed={boat.max_speed}")
    print(f"Plane: capacity={plane.capacity}, is_engine_powered={plane.is_engine_powered}, max_speed={plane.max_speed}")       
```

## Exercise 4
Write a multiple-inheritance scheme:
- `Vehicle` as above (with a small difference);
- `Engine`: a combustion engine, with a method returning the fuel type as an `Enum`
  (diesel, gasoline, kerosene, coal, electricity, solar...) and a method giving average consumption;
- `Donkey` and `Bicycle`: vehicles without engine;
- `Boat`, `Car`, `Airplane`: vehicles with engine.

```python
%%writefile code/exercise4.py
from enum import Enum

# write your solution herefrom enum import Enum

class FuelType(Enum):
    DIESEL = "diesel"
    GASOLINE = "gasoline"
    KEROSENE = "kerosene"
    COAL = "coal"
    ELECTRICITY = "electricity"
    SOLAR = "solar"

class Vehicle:
    def __init__(self, capacity, is_engine_powered, max_speed):
        self.capacity = capacity
        self.is_engine_powered = is_engine_powered
        self.max_speed = max_speed

# Engine component for vehicles with an engine
class Engine:
    def __init__(self, fuel_type: FuelType, average_consumption: float):
        self.fuel_type = fuel_type
        self.average_consumption = average_consumption  # e.g., liters per 100km or kWh per km

    def get_fuel_type(self) -> FuelType:
        return self.fuel_type

    def get_average_consumption(self) -> float:
        return self.average_consumption

# Vehicles without engine (inheriting only from Vehicle)
class Donkey(Vehicle):
    def __init__(self):
        super().__init__(capacity=2, is_engine_powered=False, max_speed=15)

class Bicycle(Vehicle):
    def __init__(self):
        super().__init__(capacity=1, is_engine_powered=False, max_speed=30)

# Vehicles with engine (using multiple inheritance: Vehicle + Engine)
class Boat(Vehicle, Engine):
    def __init__(self, fuel_type: FuelType, average_consumption: float):
        Vehicle.__init__(self, capacity=15, is_engine_powered=True, max_speed=60)
        Engine.__init__(self, fuel_type=fuel_type, average_consumption=average_consumption)

class Car(Vehicle, Engine):
    def __init__(self, fuel_type: FuelType, average_consumption: float):
        Vehicle.__init__(self, capacity=5, is_engine_powered=True, max_speed=200)
        Engine.__init__(self, fuel_type=fuel_type, average_consumption=average_consumption)

class Airplane(Vehicle, Engine):
    def __init__(self, fuel_type: FuelType, average_consumption: float):
        Vehicle.__init__(self, capacity=250, is_engine_powered=True, max_speed=900)
        Engine.__init__(self, fuel_type=fuel_type, average_consumption=average_consumption)

if __name__ == '__main__':
    my_car = Car(fuel_type=FuelType.GASOLINE, average_consumption=7.5)
    print(f"Car Capacity: {my_car.capacity}, Fuel: {my_car.get_fuel_type().value}, Consumption: {my_car.get_average_consumption()}L/100km")
    
    my_bike = Bicycle()
    print(f"Bicycle Capacity: {my_bike.capacity}, Engine Powered: {my_bike.is_engine_powered}")

    my_boat = Boat(fuel_type=FuelType.DIESEL, average_consumption=10.0)
    print(f"Boat Capacity: {my_boat.capacity}, Fuel: {my_boat.get_fuel_type().value}, Consumption: {my_boat.get_average_consumption()}L/100km")

    my_airplane = Airplane(fuel_type=FuelType.KEROSENE, average_consumption=2.5)
    print(f"Airplane Capacity: {my_airplane.capacity}, Fuel: {my_airplane.get_fuel_type().value}, Consumption: {my_airplane.get_average_consumption()}L/100km")
```

## Exercise 5
Write a `DoubleLinkedList` container with:
- a constructor of an empty list,
- an iterator (`__iter__`),
- adding an element at the beginning,
- removing the first element.

Add a test program that: creates a list, adds ~10 elements, displays it in normal then reverse order,
removes the first 5, displays it again, and verifies the removed elements are gone.

```python
%%writefile code/exercise5.py

class Node:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None

class DoubleLinkedList:
    def __init__(self,data):
        self.data = data
        self.prev = None
        self.next = None

class DoubleLinkedList:
    def __init__(self, data):
        self.head = None
        self.tail = None

    #add element at beginning
    def add_first(self,data):
        new_node = Node(data)
        if self.head is None:
            self.head = self.tail = new_node
        else:
            new_node.next = self.head
            self.head.prev = new_node
            self.head = new_node

    #remove first element
    def remove_first(self):
        if self.head is None:
            return None

        removed_data = self.head.data
        if self.head == self.tail:
            self.head = self.tail = None
        else:
            self.head = self.head.next
            self.head.prev = None

        return removed_data

    #iterator
    def __iter__(self):
        current = self.head
        while current:
            yield current.data
            current = current.next

    def reverse_iter(self):
        current = self.tail
        while current:
            yield current.data
            current = current.prev

if __name__ == "__main__":
    # add your test here
    print("Creating a new empty doublelinkedlist...")
    dll = DoubleLinkedList(0)

    #add 10 elements at the beginning
    print("\nAdding elements 1 through 10 at the beginning of the list...")
    for i in range(1, 11):
        dll.add_first(i)

    #normal order
    print("\nDisplaying in Normal order:")
    print(list(dll))
    
    #reverse order
    print("\nReverse order:")
    print(list(dll.reverse_iter()))

    #remove first 5 eelements
    print("\nRemoving first 5 elements:")
    for _ in range(5):
        dll.remove_first()

    #display again, verify the removed elements r gone
    print("\nList after remove:")
    print(list(dll))
    print("\nVerification: Removed elements 10 through 6 are gone")
     
```
